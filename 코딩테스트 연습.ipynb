```python
# 정렬

arr = [2, 1, 7, 5]
arr = [(1,3),(1,2),(2,1),(2,5)]

# 오름차순
arr = sorted(arr)
print(arr)
# 내림차순
arr = sorted(arr, reverse=True)
print(arr)

# 조건부 정렬 (예: 첫 번째 오름차순, 두 번째 내림차순)
arr = sorted(arr, key=lambda x: (x[0], -x[1]))
print(arr)
```

```python
s = "banana"

# 정렬
s = "".join(sorted(s))
print(s)


s = "banana"

# 뒤집기
s = s[::-1]
print(s)

# 카운팅
from collections import Counter
cnt = Counter(s)
cnt = cnt['b']
print(cnt)
```

```python
arr = [1, 2, 3, 4]

# 합, 최댓값, 최솟값
sum(arr), max(arr), min(arr)
print(sum(arr))
print(max(arr))
print(min(arr))

# 슬라이싱
arr[1:3]
print(arr[1:3])

# enumerate
for i, v in enumerate(arr):
    print(i, v)
```

```python
# 스택 (리스트)
stack = []
stack.append(1)
ps = stack.pop()
print(ps)

# 큐 (deque)
from collections import deque
q = deque([1, 2, 3])
q.append(4)
pq = q.popleft()
print(pq)
```

```python
s1 = {1, 2, 3}
s2 = {3, 4, 5}
print(s1 & s2) # 교집합
print(s1 | s2) # 합집합
print(s1 - s2) # 차집합
```

```python
from collections import deque

graph1 = {
    1: [2, 3],
    2: [1, 4],
    3: [1, 5],
    4: [2],
    5: [3],
}

def bfs(start, graph):
    q = deque([start])
    visited = set([start])
    while q:
        x = q.popleft()
        for nx in graph[x]:
            if nx not in visited:
                visited.add(nx)
                q.append(nx)
    return visited

bfs(1, graph1)
```
